```python
def create_padding_mask(seq, pad_idx):
    """
    Tạo mask cho key_padding_mask (True là Pad).
    Shape: (Batch_Size, Seq_Len)
    """
    return (seq == pad_idx)
```

```python
from torch import nn 
import torch.nn.functional as F
class P_gen(nn.Module):
            def __init__(self, d_model = 300):
                super().__init__()
                self.W_ht = nn.Linear(d_model\
                    , d_model)
                self.W_st = nn.Linear(d_model\
                    , d_model)
                self. W_xt = nn.Linear(d_model\
                    , d_model)
            def forward(self, x_t, s_t, h_t): 
                scores = self.W_ht(h_t) + \
                    self.W_st(s_t) + self.W_xt(x_t) 
                return F.sigmoid(scores)
```

## ***P_gen.shape(): (B, 1, 1)***

```python
P_generator = P_gen()
```

```python
batch_size = 4
sequence_length = 100
d_model = 300

x_t = torch.randn(batch_size, sequence_length, d_model)
s_t = torch.randn(batch_size, sequence_length, d_model)
h_t = torch.randn(batch_size, sequence_length, d_model)
```

```python
tensors  = [x_t, s_t, h_t] 
```

```python
result = P_generator(x_t = x_t, s_t = s_t, h_t = h_t)
```
